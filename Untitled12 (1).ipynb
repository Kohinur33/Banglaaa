```python
import os
import glob
from datetime import datetime

search_directories = [
    os.path.join(os.path.expanduser('~'), 'Downloads'),
    os.path.join(os.path.expanduser('~'), 'Desktop'),
    r'E:\License Plate',
    r'E:\\',
    r'C:\\'
]

print("রিসেন্ট ডাউনলোড হওয়া জিপ ও ডেটাসেট ফোল্ডার খোঁজা হচ্ছে...\n")

found_files = []

for base_dir in search_directories:
    if not os.path.exists(base_dir):
        continue
    try:
        for root, dirs, files in os.walk(base_dir):
            if '.venv' in root or '$Recycle.Bin' in root or 'System Volume Information' in root:
                continue
            for f in files:
                if f.endswith(('.zip', '.rar', '.7z', '.tar', '.gz')):
                    full_path = os.path.join(root, f)
                    try:
                        mtime = os.path.getmtime(full_path)
                        found_files.append((full_path, mtime))
                    except Exception:
                        pass
    except Exception:
        pass

# সময় অনুযায়ী সর্ট করা (সবচেয়ে নতুন ফাইলটি সবার আগে দেখাবে)
found_files.sort(key=lambda x: x[1], reverse=True)

if found_files:
    print("খুঁজে পাওয়া নতুন আর্কাইভ/জিপ ফাইলসমূহ:\n")
    for file_path, mtime in found_files[:10]: # প্রথম ১০টি রিসেন্ট ফাইল দেখাবে
        time_str = datetime.fromtimestamp(mtime).strftime('%Y-%m-%d %H:%M:%S')
        print(f"📁 পাথ: {file_path}")
        print(f"   ডাউনলোডের সময়: {time_str}\n")
else:
    print("কোনো .zip বা আর্কাইভ ফাইল পাওয়া যায়নি।")
```

```python
import os
import shutil
import random

source_blurred = "/content/deblur_dataset/blurred"
source_original = "/content/deblur_dataset/original"

base_dir = "/content/deblur_data"

for folder in [
    "train/blurred", "train/original",
    "val/blurred", "val/original"
]:
    os.makedirs(os.path.join(base_dir, folder), exist_ok=True)

files = [
    f for f in os.listdir(source_blurred)
    if os.path.isfile(os.path.join(source_blurred, f))
]

random.seed(42)
random.shuffle(files)

split = int(len(files) * 0.8)

train_files = files[:split]
val_files = files[split:]

for filename in train_files:
    shutil.copy2(
        os.path.join(source_blurred, filename),
        os.path.join(base_dir, "train/blurred", filename)
    )
    shutil.copy2(
        os.path.join(source_original, filename),
        os.path.join(base_dir, "train/original", filename)
    )

for filename in val_files:
    shutil.copy2(
        os.path.join(source_blurred, filename),
        os.path.join(base_dir, "val/blurred", filename)
    )
    shutil.copy2(
        os.path.join(source_original, filename),
        os.path.join(base_dir, "val/original", filename)
    )

print("Deblur train/validation split complete!")
print("Train pairs:", len(train_files))
print("Validation pairs:", len(val_files))
```

```python
import torch
import torch.nn as nn

class DeblurAutoencoder(nn.Module):
    def __init__(self):
        super().__init__()

        # Encoder
        self.encoder = nn.Sequential(
            nn.Conv2d(3, 64, 3, padding=1),
            nn.ReLU(),
            nn.Conv2d(64, 128, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )

        # Decoder
        self.decoder = nn.Sequential(
            nn.Conv2d(128, 64, 3, padding=1),
            nn.ReLU(),
            nn.Upsample(scale_factor=2, mode="bilinear", align_corners=False),
            nn.Conv2d(64, 3, 3, padding=1),
            nn.Sigmoid()
        )

    def forward(self, x):
        x = self.encoder(x)
        x = self.decoder(x)
        return x


device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = DeblurAutoencoder().to(device)

print("Deblur model created successfully!")
print("Device:", device)
```

```python
from torch.utils.data import Dataset, DataLoader
from PIL import Image
import os
import glob
import torchvision.transforms as transforms

class DeblurDataset(Dataset):
    def __init__(self, blurred_dir, original_dir):
        self.blurred_dir = blurred_dir
        self.original_dir = original_dir

        self.files = [
            os.path.basename(f)
            for f in glob.glob(os.path.join(blurred_dir, "*"))
        ]

        self.transform = transforms.Compose([
            transforms.Resize((256, 256)),
            transforms.ToTensor()
        ])

    def __len__(self):
        return len(self.files)

    def __getitem__(self, idx):
        filename = self.files[idx]

        blurred = Image.open(
            os.path.join(self.blurred_dir, filename)
        ).convert("RGB")

        original = Image.open(
            os.path.join(self.original_dir, filename)
        ).convert("RGB")

        blurred = self.transform(blurred)
        original = self.transform(original)

        return blurred, original


train_dataset = DeblurDataset(
    "/content/deblur_data/train/blurred",
    "/content/deblur_data/train/original"
)

val_dataset = DeblurDataset(
    "/content/deblur_data/val/blurred",
    "/content/deblur_data/val/original"
)

train_loader = DataLoader(
    train_dataset,
    batch_size=16,
    shuffle=True,
    num_workers=2
)

val_loader = DataLoader(
    val_dataset,
    batch_size=16,
    shuffle=False,
    num_workers=2
)

print("Train pairs:", len(train_dataset))
print("Validation pairs:", len(val_dataset))
print("DataLoader ready!")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

epochs = 10

for epoch in range(epochs):

    model.train()
    train_loss = 0.0

    for blurred, original in train_loader:
        blurred = blurred.to(device)
        original = original.to(device)

        optimizer.zero_grad()

        output = model(blurred)
        loss = criterion(output, original)

        loss.backward()
        optimizer.step()

        train_loss += loss.item()

    # Validation
    model.eval()
    val_loss = 0.0

    with torch.no_grad():
        for blurred, original in val_loader:
            blurred = blurred.to(device)
            original = original.to(device)

            output = model(blurred)
            loss = criterion(output, original)

            val_loss += loss.item()

    train_loss /= len(train_loader)
    val_loss /= len(val_loader)

    print(
        f"Epoch [{epoch+1}/{epochs}] "
        f"Train Loss: {train_loss:.6f} "
        f"Val Loss: {val_loss:.6f}"
    )

print("Deblurring training complete!")
```

```python
import torch
import matplotlib.pyplot as plt
from PIL import Image
import torchvision.transforms as transforms

# একটি blurred test image
test_file = "/content/blurred_dataset/test/images/" + \
            os.listdir("/content/blurred_dataset/test/images")[0]

image = Image.open(test_file).convert("RGB")

transform = transforms.Compose([
    transforms.Resize((256, 256)),
    transforms.ToTensor()
])

input_image = transform(image).unsqueeze(0).to(device)

model.eval()

with torch.no_grad():
    output = model(input_image)

output = output.squeeze(0).cpu().permute(1, 2, 0).numpy()

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(image.resize((256, 256)))
plt.title("Blurred Input")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(output)
plt.title("Deblurred Output")
plt.axis("off")

plt.show()
```
